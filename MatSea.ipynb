```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df=Student_marks=pd.read_csv("Student_marks.csv")
df
```

```python
fig, ax = plt.subplots()
ax.plot(df['time_study'], df['Marks'])
ax.set_xlabel('time_study')
ax.set_ylabel('Marks')
plt.title('Study Time vs Marks')
```

```python
# Sort by study time so the line plot is ordered on the x-axis
df_line = df.sort_values('time_study')

fig, ax = plt.subplots()
ax.plot(df_line['time_study'], df_line['Marks'], marker='o', linestyle='-')
ax.set_xlabel('time_study')
ax.set_ylabel('Marks')
ax.set_title('Line Plot: Study Time vs Marks')
plt.show()
```

```python
# Histogram: distribution of marks
fig, ax = plt.subplots()
ax.hist(df['Marks'], bins=10, color='orange', edgecolor='black')
ax.set_xlabel('Marks')
ax.set_ylabel('Frequency')
ax.set_title('Histogram of Marks')
plt.show()
```

```python
# Bar graph: average marks by number of courses
avg_marks = df.groupby('number_courses', as_index=False)['Marks'].mean()

fig, ax = plt.subplots()
ax.bar(avg_marks['number_courses'], avg_marks['Marks'], color='teal', edgecolor='black')
ax.set_xlabel('number_courses')
ax.set_ylabel('Average Marks')
ax.set_title('Average Marks by Number of Courses')
plt.show()
```

```python
# Scatter plot: study time vs marks
fig, ax = plt.subplots()
ax.scatter(df['time_study'], df['Marks'], color='crimson', edgecolor='black', alpha=0.8)
ax.set_xlabel('time_study')
ax.set_ylabel('Marks')
ax.set_title('Scatter Plot: Study Time vs Marks')
plt.show()
```
